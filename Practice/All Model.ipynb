```python
import numpy as np 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt 
import warnings

warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('titanic')
df.head(15)
```

```python
df.columns
```

```python
df.info()
```

```python
df.drop(['embark_town', 'class', 'adult_male', 'deck',
       'alive','who'], axis=1,inplace=True)
```

```python
df.columns
```

```python
df["age"].fillna(df["age"].mean(),inplace=True)
```

```python
df.dropna(subset=["embarked"],inplace=True)
```

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

```

```python
df.head()
```

```python
df['sex'] = le.fit_transform(df['sex'])
df["embarked"] = le.fit_transform(df["embarked"])
```

```python
df.astype(int)
```

```python
x =  df.drop("survived", axis=1)
y = df["survived"]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.20,random_state=42)
```

***LOGISTIC_REGRESSION MODEL***

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
accuracy_score(y_test,y_pred)
```

```python
confusion_matrix(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
x_train_scaled
```

```python
x_test_scaled = scaler.fit_transform(x_test)
x_test_scaled
```

***KNN MODEL***

```python
from sklearn.neighbors import KNeighborsClassifier
KNN_model = KNeighborsClassifier(n_neighbors = 5)
KNN_model.fit(x_train_scaled,y_train)
```

```python
y_pred_KNN = KNN_model.predict(x_test_scaled)
y_pred_KNN
```

```python
accuracy_score(y_test,y_pred_KNN)
```

```python
confusion_matrix(y_test,y_pred_KNN)
```

```python
print(classification_report(y_test,y_pred_KNN))
```

***NAIVE_BAYES MODEL***

```python
from sklearn.naive_bayes import GaussianNB
model_NB = GaussianNB()
```

```python
model_NB.fit(x_train,y_train)
y_pred_NB = model_NB.predict(x_test)
```

```python
y_pred_NB
```

```python
y_test
```

```python
accuracy_score(y_pred_NB,y_test)
```

```python
confusion_matrix(y_pred_NB,y_test)
```

```python
print(classification_report(y_pred_NB,y_test))
```

***DECISION_TREE MODEL***

```python
from sklearn.tree import DecisionTreeClassifier
model_DT = DecisionTreeClassifier(random_state=42)
```

```python
model_DT.fit(x_train_scaled,y_train)
```

```python
y_pred_DT = model_DT.predict(x_test_scaled)
```

```python
y_pred_DT
```

```python
y_test
```

```python
accuracy_score(y_pred_DT,y_test)
```

```python
confusion_matrix(y_pred_DT,y_test)
```

```python
print(classification_report(y_pred_DT,y_test))
```

***SVM MODEL***

```python
from sklearn.svm import SVC
model_SVM = SVC(kernel="linear")
```

```python
model_SVM.fit(x_train_scaled,y_train)
```

```python
y_pred_svc = model_SVM.predict(x_test_scaled)
```

```python
accuracy_score(y_pred_svc,y_test)
```

```python
confusion_matrix(y_pred_svc,y_test)
```

```python
print(classification_report(y_pred_svc,y_test))
```
